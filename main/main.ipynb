```python
# Force reload modules
import importlib
import sys

# Remove cached modules
for module in list(sys.modules.keys()):
    if module.startswith('utils') or module.startswith('model') or module.startswith('gui'):
        del sys.modules[module]

import sys
import os
sys.path.append(os.path.abspath(".")) 

import tkinter as tk
from model import DataLoader, DataPreprocessor, ArimaConfig, ArimaModel
from utils import DataValidator, CurrencyConverter, PredictionService
from gui import MainWindow, PredictionFormatter

class GoldPriceApp:
    def __init__(self):
        self._setup_model()
        self._setup_gui()

    def _setup_model(self):
        loader = DataLoader(r"C:\Users\HP\Documents\GoldPricePrediction\GoldPriceDataset.csv")
        df = loader.load()

        preprocessor = DataPreprocessor()
        series = preprocessor.preprocess(df)

        # Fixed: pass individual parameters, not positional
        self.model = ArimaModel(
            ArimaConfig(p=1, d=1, q=1)
        )
        self.model.train(series)

        self.prediction_service = PredictionService(self.model)

    def _setup_gui(self):
        self.root = tk.Tk()
        self.window = MainWindow(self.root, self._predict)

    def _predict(self, year):
        try:
            print(f"Step 1: Validating year {year}")
            DataValidator.validate_year(year)
        
            print(f"Step 2: Calling predict_year({year})")
            predictions = self.prediction_service.predict_year(year)
        
            print(f"Step 3: Type of predictions: {type(predictions)}")
            print(f"Step 4: Length of predictions: {len(predictions)}")
            print(f"Step 5: First 3 values: {predictions[:3]}")
            print(f"Step 6: Predictions is a list? {isinstance(predictions, list)}")
        
            # Force convert to list
            predictions_list = list(predictions)
            print(f"Step 7: After converting to list, length: {len(predictions_list)}")
        
            print(f"Step 8: Formatting...")
            result = PredictionFormatter.format(predictions_list, CurrencyConverter)
        
            print(f"Step 9: Success!")
            return result
        except Exception as e:
            import traceback
            error_details = traceback.format_exc()
            print(error_details)
            return f"Error: {e}\n\n{error_details}"

    def run(self):
        self.root.mainloop()


if __name__ == "__main__":
    app = GoldPriceApp()
    app.run()
```
